# Objetivo
O objetivo desse projeto é realizar a análise exploratória dos dados do dataset do Titanic e construir um modelo preditivo, utilizando técnicas de machine learning para prever a sobrevivência dos passageiros com base em suas características. 

O projeto inclui etapas de limpeza e preparação dos dados, análise exploratória, construção e avaliação de modelos preditivos, e interpretação dos resultados utilizando métodos como SHAP para entender a importância das features.
